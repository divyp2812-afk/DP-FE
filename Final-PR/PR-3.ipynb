6. Handle missing data with:

````python
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.impute import KNNImputer
from sklearn.impute import SimpleImputer

# 1. Simple Imputer – Numerical (Mean/Median)
df_most = Data.copy()
imputer = SimpleImputer(strategy="mean")
Data["annual_income"] = imputer.fit_transform(Data[["annual_income"]])
print(Data.head())

df_most = Data.copy()   
imputer = SimpleImputer(strategy="median")
Data["loan_amount"] = imputer.fit_transform(Data[["loan_amount"]])
print(Data.head())

# 2. Simple Imputer (Categorical: Most Frequent)
categorical_cols = Data.select_dtypes(include=["object", "category"]).columns
imputer = SimpleImputer(strategy="most_frequent")
Data[categorical_cols] = imputer.fit_transform(Data[categorical_cols])
print("Missing values after categorical imputation:")
print(Data[categorical_cols].isnull().sum())

# 3. Most Frequent Category Imputation
Data["gender"] = Data["gender"].fillna(Data["gender"].mode()[0])
print(Data["gender"].isnull().sum())

# 4. Missing Indicator + Random Sample Imputation

imputer = IterativeImputer(random_state=42)
Data[["annual_income"]] = imputer.fit_transform(Data[["annual_income"]])
print(Data.head())

# 5. KNN Imputer (Multivariate)
imputer = KNNImputer(n_neighbors=5)
Data[["annual_income"]] = imputer.fit_transform(Data[["annual_income"]])
print(Data.head())

# 6. MICE Algorithm
imputer = IterativeImputer(random_state=42)
Data[["annual_income"]] = imputer.fit_transform(Data[["annual_income"]])
print(Data.head())

# 7. Complete Case Analysis(dropping rows/columns).
Data.dropna(inplace=True)
print(Data.head())
````

## Part-D: Outlier Handling

7. Detect and treat outliers using: 

````python
# Z-score method
from scipy import stats

z_scores = stats.zscore(Data["annual_income"])
Data = Data[abs(z_scores) < 3]
print("Z-score", Data.head())

# IQR method
Q1 = Data["loan_amount"].quantile(0.25)
Q3 = Data["loan_amount"].quantile(0.75)
IQR = Q3 - Q1
Data = Data[(Data["loan_amount"] >= Q1 - 1.5 * IQR) & (Data["loan_amount"] <= Q3 + 1.5 * IQR)]
print("IQR", Data.head())

# Percentile method
lower_percentile = Data["loan_amount"].quantile(0.01)
upper_percentile = Data["loan_amount"].quantile(0.99)
Data = Data[(Data["loan_amount"] >= lower_percentile) & (Data["loan_amount"] <= upper_percentile)]
print("Percentile", Data.head())

# Winsorization method
from scipy.stats.mstats import winsorize

Data["loan_amount"] = winsorize(Data["loan_amount"], limits=[0.01, 0.01])
print("Winsorization", Data.head())
````

## Part-E: Feature Engineering

8. Handle variable types:

````python
# Mixed variable (numeric + categorical)
Data["age_group"] = pd.cut(Data["age"],bins=[18, 30, 45, 60, 100],labels=["Young", "Adult", "Middle", "Senior"])

Data["region_age"] = (Data["region"].astype(str) + "_" +Data["age_group"].astype(str))

print(Data[["region", "age", "region_age"]].head())
````

````python
# Date & Time variable
Data["join_date"] = pd.to_datetime(Data["join_date"], errors="coerce")

Data["year"] = Data["join_date"].dt.year
Data["month"] = Data["join_date"].dt.month
Data["day"] = Data["join_date"].dt.day
Data["weekday"] = Data["join_date"].dt.weekday

print(Data[["join_date", "year", "month", "day", "weekday"]].head())
````

9. Encoding categorical variables:

````python
# Ordinal Encoding
Data["education_encoded"] = Data["education_level"].astype("category").cat.codes
print(Data[["education_level", "education_encoded"]].head())
````

````python
# Label Encoding
Data["gender_encoded"] = Data["gender"].astype("category").cat.codes
print(Data[["gender", "gender_encoded"]].head())
````

````python
# One-Hot Encoding
cat_cols = Data.select_dtypes(include=["object", "category"]).columns

Data = pd.get_dummies(Data, columns=cat_cols, drop_first=True)

print(Data.head())
````

10. Encoding numerical features:

````python
# Binning(discretize income into groups)
Data["income_binned"] = pd.cut(Data["annual_income"], bins=5, labels=False)
print(Data[["annual_income", "income_binned"]].head())
````

````python
# Binarization (flag if>threshold)
Data["income_flag"] = (Data["annual_income"] > Data["annual_income"].median()).astype(int)
print(Data[["annual_income", "income_flag"]].head())
````

````python
# Quantile Binning
Data["income_quantile"] = pd.qcut(Data["annual_income"], q=5, labels=False, duplicates="drop")
print(Data[["annual_income", "income_quantile"]].head())
````

````python
# K-Means Binning
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=5, random_state=42)
Data["income_kmeans"] = kmeans.fit_predict(Data[["annual_income"]])
print(Data[["annual_income", "income_kmeans"]].head())
````

## Part-G: Feature Construction & Transformation

12. Apply transformations:

````python
# FunctionTransformer (Log Transformation,reciprocal,square root)
from sklearn.preprocessing import FunctionTransformer
import numpy as np
log_transformer = FunctionTransformer(func=np.log1p, inverse_func=np.expm1)
Data["annual_income_log"] = log_transformer.fit_transform(Data[["annual_income"]])
reciprocal_transformer = FunctionTransformer(func=np.reciprocal, inverse_func=np.reciprocal)
Data["annual_income_reciprocal"] = reciprocal_transformer.fit_transform(Data[["annual_income"]])
sqrt_transformer = FunctionTransformer(func=np.sqrt, inverse_func=np.square)
Data["annual_income_sqrt"] = sqrt_transformer.fit_transform(Data[["annual_income"]])
print(Data[["annual_income", "annual_income_log", "annual_income_reciprocal", "annual_income_sqrt"]].head())
````

````python
# PowerTransformer (Box-Cox Transformation,Yeo-Johnson Transformation)
from sklearn.preprocessing import PowerTransformer

box_cox_transformer = PowerTransformer(method='box-cox')
Data["annual_income_boxcox"] = box_cox_transformer.fit_transform(Data[["annual_income"]])

yeo_johnson_transformer = PowerTransformer(method='yeo-johnson')
Data["annual_income_yeojohnson"] = yeo_johnson_transformer.fit_transform(Data[["annual_income"]])
print(Data[["annual_income", "annual_income_boxcox", "annual_income_yeojohnson"]].head())
````

````python
# Column Transformer (apply different preprocessing steps to different columns).
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import PowerTransformer

column_transformer = ColumnTransformer(
    transformers=[
        ('boxcox', PowerTransformer(method='box-cox'), ['annual_income']),
        ('yeojohnson', PowerTransformer(method='yeo-johnson'), ['annual_income'])
    ]
)

df = column_transformer.fit_transform(Data)
print(df)
````
