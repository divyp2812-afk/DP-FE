9. Deploy the Model - 
Make the model available for real-world use.

Update and maintain the model over time.

# 4. Tensors

## What are Tensors?

A **Tensor** is a multi-dimensional array used to store and process numerical data. It is a generalization of a scalar (0D), vector (1D), and matrix (2D). Tensors are widely used in **Machine Learning**, **Deep Learning**, and **Data Science** for representing and manipulating data efficiently.

---

## In-depth Explanation of Tensors with Simple Examples Using NumPy

### 1. Scalar (0D Tensor)
A scalar is a tensor that contains only one value.

```python
import numpy as np

scalar = np.array(10)
print(scalar)
```

**Output:**

```
10
```

---

### 2. Vector (1D Tensor)
A vector is a one-dimensional tensor containing multiple values.

```python
import numpy as np

vector = np.array([10, 20, 30])
print(vector)
```

**Output:**

```
[10 20 30]
```

---

### 3. Matrix (2D Tensor)
A matrix is a two-dimensional tensor with rows and columns.

```python
import numpy as np

matrix = np.array([[1, 2],
                   [3, 4]])

print(matrix)
```

**Output:**

```
[[1 2]
 [3 4]]
```

---

### 4. 3D Tensor
A 3D tensor is a collection of matrices.

```python
import numpy as np

tensor = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print(tensor)
```

**Output:**

```
[[[1 2]
  [3 4]]

 [[5 6]
  [7 8]]]
```

---

### Tensor Properties

```python
print("Shape:", tensor.shape)
print("Dimensions:", tensor.ndim)
print("Size:", tensor.size)
```

**Output:**

```
Shape: (2, 2, 2)
Dimensions: 3
Size: 8
```

### 7. Apply data cleaning:
         - Handle missing data(imputation or removal).
         - Correct inconsistent data types.
         - Drop irrelevant columns.

````python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Income"] = df["Income"].fillna(df["Income"].mean())
df["Purchases"] = df["Purchases"].fillna(df["Purchases"].mean())
print(df.isnull().sum())
````

````python
df["Age"] = df["Age"].astype(int)
df["Income"] = df["Income"].astype(float)
df["Purchases"] = df["Purchases"].astype(int)
print(df.dtypes)
````

````python
df.drop(columns=["Customer_ID", "Name", "City", "Phone"], inplace=True)
print(df.columns)
````

# Part-D: Exploratory Data Analysis(EDA)

### 8. Perform Univariate Analysis:
         - Distribution plots of Age, Income and Purchases.

````python
import matplotlib.pyplot as plt

# Distribution plot of Age

plt.figure(figsize=(15, 5))
plt.hist(df["Age"], bins=10)
plt.title("Distribution of Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()

# Distribution plot of Income

plt.figure(figsize=(15, 5))
plt.hist(df["Income"], bins=10)
plt.title("Distribution of Income")
plt.xlabel("Income")
plt.ylabel("Frequency")
plt.show()

# Distribution plot of Purchases

plt.figure(figsize=(15, 5))
plt.hist(df["Purchases"], bins=10)
plt.title("Distribution of Purchases")
plt.xlabel("Purchases")
plt.ylabel("Frequency")
plt.show()
````

### 9. Perform Bivariate Analysis:
         - Relationship between Gender & Purchases.
         - Relationship between Income & Churn.

````python
import seaborn as sns

plt.figure(figsize=(15, 5))
sns.boxplot(x="Income", y="Churn", data=df)
plt.title("Relationship between Income and Churn")
plt.xlabel("Income")
plt.ylabel("Churn")
plt.show()
````

````python
plt.figure(figsize=(6,4))
sns.barplot(data=df, x="Gender", y="Purchases")
plt.title("Relationship between Gender and Purchases")
plt.xlabel("Gender")
plt.ylabel("Purchases")
plt.show()
````

### 10. Perform Multivariate Analysis:
         - Correlation heatmap of all numerical variables.
         - Pair plots to identify feature interactions.

````python
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
````

````python
sns.pairplot(df[['Age', 'Income', 'Purchases']])
plt.show()
````
